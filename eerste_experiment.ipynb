## Doel

In dit experiment onderzoek ik wat er met het gemiddelde van een dataset
gebeurt wanneer ik een extreem hoge waarde toevoeg.

### Verwachting

Ik verwacht dat het gemiddelde hoger wordt door de extreme waarde.


## Arrange

Eerst maak ik een eenvoudige dataset met vijf waarden.
Deze gegevens vormen de beginsituatie van het experiment.


```python
data = [6, 7, 7, 8, 7]

print(data)
```

```python
gemiddelde = sum(data) / len(data)

print(gemiddelde)
```

## Act

Nu voeg ik een extreme waarde van 20 toe aan de dataset.
Daarna bereken ik het gemiddelde opnieuw.


```python
nieuwe_data = data.copy()
nieuwe_data.append(20)

nieuw_gemiddelde = sum(nieuwe_data) / len(nieuwe_data)

print(nieuwe_data)
print(nieuw_gemiddelde)
```

## Assert
Heeft het experiment laten zien wat ik wilde onderzoeken en klopt het met mijn verwachting?

Het oorspronkelijke gemiddelde was 7,0.

Na het toevoegen van de extreme waarde 20 is het gemiddelde hoger geworden.

Mijn verwachting was daarom correct. De extreme waarde heeft duidelijk
invloed op het gemiddelde.
