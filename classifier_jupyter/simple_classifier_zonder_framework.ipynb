# Eenvoudige Machine Learning Classifier

In deze notebook bouwen we een eenvoudige machine-learningclassifier zonder gebruik te maken van een ML-framework.

Het doel is om de basisprincipes van machine learning te begrijpen voordat we dezelfde principes toepassen met PyTorch.

Het model gebruikt twee features:

- lichaamslengte
- lichaamsgewicht

Het model voorspelt één van twee klassen: **klasse 0** of **klasse 1**.

Tijdens het trainen leert het model zelf de **weights** en **bias**.

## 1. Trainingsdata

De trainingsdata bestaat uit voorbeelden waarvan het juiste antwoord al bekend is.

Ieder trainingsvoorbeeld bevat:

1. lichaamslengte - feature
2. lichaamsgewicht - feature
3. juiste klasse - label

Het model gebruikt deze voorbeelden tijdens de training om de weights en bias aan te passen.

```python
training_data = [
    [160, 55, 0],
    [165, 60, 0],
    [170, 65, 0],
    [180, 80, 1],
    [185, 85, 1],
    [190, 90, 1]
]

training_data
```

## 2. Modelparameters

Het model gebruikt twee weights en één bias.

De weights bepalen hoeveel invloed de twee features hebben op de berekening.  
De bias verschuift de beslissingsgrens van het model.

Voor de training beginnen de weights en bias op '0.0'.

De 'learning_rate' bepaalt hoe groot de aanpassingen tijdens het leren zijn.

```python
ml_weight_lichaamslengte = 0.0
ml_weight_lichaamsgewicht = 0.0
ml_bias = 0.0

learning_rate = 0.001

print("ML-weight lichaamslengte:", ml_weight_lichaamslengte)
print("ML-weight lichaamsgewicht:", ml_weight_lichaamsgewicht)
print("ML-bias:", ml_bias)
print("Learning rate:", learning_rate)
```

## 3. Voorspelling

De functie `voorspel()` berekent eerst een ML-waarde:

**ml_waarde = lichaamslengte × weight_lengte + lichaamsgewicht × weight_gewicht + bias**

Daarna wordt de ML-waarde omgezet naar een klasse:

- ml_waarde ≥ 0 → klasse 1
- ml_waarde < 0 → klasse 0

Dezelfde functie wordt zowel tijdens training als later tijdens inference en validatie gebruikt.

```python
def voorspel(lichaamslengte, lichaamsgewicht):
    ml_waarde = (
        lichaamslengte * ml_weight_lichaamslengte
        + lichaamsgewicht * ml_weight_lichaamsgewicht
        + ml_bias
    )

    if ml_waarde >= 0:
        voorspelling = 1
    else:
        voorspelling = 0

    return voorspelling, ml_waarde
```

## 4. Ongetraind model testen

Voordat het model wordt getraind, voeren we één voorspelling uit.

De weights en bias zijn op dit moment nog '0.0'. Het model heeft dus nog niets geleerd.

Hiermee kunnen we later het verschil tussen het ongetrainde en het getrainde model zien.

```python
voorspelling, ml_waarde = voorspel(160, 55)

print("ML-waarde:", ml_waarde)
print("Voorspelling:", voorspelling)
```

## 5. Training

Tijdens de training doorloopt het model meerdere keren alle trainingsvoorbeelden.

Eén volledige ronde door de trainingsdata noemen we een **epoch**.

Per trainingsvoorbeeld:

1. maakt het model een voorspelling;
2. wordt de voorspelling vergeleken met het juiste antwoord;
3. wordt de fout berekend;
4. worden bij een fout de weights en bias aangepast.

Na meerdere epochs ontstaat een getraind model.

```python
aantal_epochs = 10

for epoch in range(aantal_epochs):
    aantal_fouten = 0

    for trainingsvoorbeeld in training_data:
        lichaamslengte = trainingsvoorbeeld[0]
        lichaamsgewicht = trainingsvoorbeeld[1]
        juiste_antwoord = trainingsvoorbeeld[2]

        voorspelling, ml_waarde = voorspel(
            lichaamslengte,
            lichaamsgewicht
        )

        fout = juiste_antwoord - voorspelling

        if fout != 0:
            aantal_fouten = aantal_fouten + 1

            ml_weight_lichaamslengte = (
                ml_weight_lichaamslengte
                + fout * lichaamslengte * learning_rate
            )

            ml_weight_lichaamsgewicht = (
                ml_weight_lichaamsgewicht
                + fout * lichaamsgewicht * learning_rate
            )

            ml_bias = (
                ml_bias
                + fout * learning_rate
            )

    print("Epoch:", epoch + 1, "- aantal fouten:", aantal_fouten)
```

## 6. Het getrainde model

Na de training bekijken we de geleerde weights en bias.

Deze waarden vormen samen de geleerde parameters van het model. Ze zijn tijdens de training ontstaan uit de trainingsdata.

```python
print("Getraind model:")
print("ML-weight lichaamslengte:", ml_weight_lichaamslengte)
print("ML-weight lichaamsgewicht:", ml_weight_lichaamsgewicht)
print("ML-bias:", ml_bias)
```

## 7. Inference

Bij **inference** gebruiken we het getrainde model om voorspellingen te maken voor nieuwe data.

Deze nieuwe voorbeelden zijn niet gebruikt tijdens de training.

De juiste klasse is hier niet opgegeven: het is juist de taak van het model om deze te voorspellen.

Tijdens inference worden de weights en bias niet meer aangepast.

```python
# INFERENCE
nieuwe_data = [
    [172, 68],
    [175, 72],
    [178, 76],
    [182, 82],
    [188, 88]
]

for nieuwe_persoon in nieuwe_data:
    lichaamslengte = nieuwe_persoon[0]
    lichaamsgewicht = nieuwe_persoon[1]

    voorspelling, ml_waarde = voorspel(
        lichaamslengte,
        lichaamsgewicht
    )

    print()
    print("Nieuwe persoon:", nieuwe_persoon)
    print("ML-waarde:", ml_waarde)
    print("Voorspelde klasse:", voorspelling)
```

## 8. Validatie

Met validatiedata controleren we hoe het getrainde model presteert op voorbeelden die niet voor de training zijn gebruikt.

In tegenstelling tot inference kennen we bij validatie het juiste antwoord wel.

We vergelijken:

**voorspelling ↔ juiste antwoord**

De weights en bias worden tijdens deze controle niet aangepast.

De accuracy geeft aan welk percentage van de validatievoorbeelden correct is voorspeld.

```python
# VALIDATIE
validation_data = [
    [173, 69, 0],
    [176, 73, 0],
    [179, 77, 1],
    [183, 83, 1],
    [187, 87, 1]
]

aantal_correct = 0

for validatievoorbeeld in validation_data:
    lichaamslengte = validatievoorbeeld[0]
    lichaamsgewicht = validatievoorbeeld[1]
    juiste_antwoord = validatievoorbeeld[2]

    voorspelling, ml_waarde = voorspel(
        lichaamslengte,
        lichaamsgewicht
    )

    if voorspelling == juiste_antwoord:
        aantal_correct = aantal_correct + 1

accuracy = aantal_correct / len(validation_data) * 100

print("Aantal correct:", aantal_correct)
print("Totaal aantal validatievoorbeelden:", len(validation_data))
print("Accuracy:", accuracy, "%")
```
